## Problem 1: Sorting and Counting Inversions
Given an array of $n$ integers, an inversion is a pair $(i, j)$ such that $i < j$ and $A[i] > A[j]$.
- **Divide and Conquer:** Enhanced Merge Sort that sorts and counts inversions simultaneously in $O(n \log n)$ time.
- **Brute Force:** Examines all pairs $(i, j)$ in $O(n^2)$ time.
- **Comparisons:** Compared against Quick Sort and Insertion Sort.

````python
# 1. Divide and Conquer: Merge Sort and Count Inversions
def count_inversions_dc(arr):
    if len(arr) <= 1:
        return arr, 0
    mid = len(arr) // 2
    left, inv_left = count_inversions_dc(arr[:mid])
    right, inv_right = count_inversions_dc(arr[mid:])
    
    merged = []
    i = j = inv_split = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            inv_split += len(left) - i
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged, inv_left + inv_right + inv_split

# 2. Brute Force Inversion Counting
def count_inversions_bf(arr):
    inv = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] > arr[j]:
                inv += 1
    return inv

# 3. Baseline Sorting Algorithms for Comparison
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + mid + quick_sort(right)

def insertion_sort(arr):
    a = arr.copy()
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a
````

````python
# Verification on Sample Input: [8, 4, 2, 1]
sample = [8, 4, 2, 1]
sorted_sample, inv_dc = count_inversions_dc(sample)
inv_bf = count_inversions_bf(sample)

print(f"Original Array: {sample}")
print(f"Divide & Conquer: Sorted = {sorted_sample}, Inversions = {inv_dc}")
print(f"Brute Force:      Inversions = {inv_bf}")
````

````python
import time, random
import matplotlib.pyplot as plt

# Performance Benchmarking: Problem 1
sizes = [100, 200, 400, 800, 1600, 3200]
t_dc, t_bf, t_quick, t_insert = [], [], [], []

for n in sizes:
    arr = [random.randint(1, 100000) for _ in range(n)]
    
    t0 = time.perf_counter()
    count_inversions_dc(arr)
    t_dc.append(time.perf_counter() - t0)
    
    t0 = time.perf_counter()
    count_inversions_bf(arr)
    t_bf.append(time.perf_counter() - t0)
    
    t0 = time.perf_counter()
    quick_sort(arr)
    t_quick.append(time.perf_counter() - t0)
    
    t0 = time.perf_counter()
    insertion_sort(arr)
    t_insert.append(time.perf_counter() - t0)

# Plotting Input Size vs Execution Time
plt.figure(figsize=(9, 5))
plt.plot(sizes, t_dc, marker='o', label='D&C Inversion Count (O(n log n))')
plt.plot(sizes, t_bf, marker='s', label='Brute Force Inversion Count (O(n²))')
plt.plot(sizes, t_quick, marker='^', label='Quick Sort (O(n log n))')
plt.plot(sizes, t_insert, marker='d', label='Insertion Sort (O(n²))')

plt.title('Problem 1: Input Size vs Execution Time')
plt.xlabel('Input Size (n)')
plt.ylabel('Execution Time (seconds)')
plt.legend()
plt.grid(True)
plt.show()
````

### Problem 1: Complexity and Experimental Analysis

1. **Time and Auxiliary Space Complexity:**
   - **Divide and Conquer (Merge Sort + Inversion Count):**
     - **Recurrence:** $T(n) = 2T(n/2) + \Theta(n)$
     - By Master Theorem ($a=2, b=2, d=1$ where $\log_2 2 = 1 = d$): $T(n) = \Theta(n \log n)$.
     - **Auxiliary Space:** $\Theta(n)$ required for storing subarrays during the merge step.
   - **Brute Force:**
     - Checks all pairs $(i, j)$ with $i < j$: Total comparisons = $\sum_{i=1}^{n-1} (n - i) = \frac{n(n-1)}{2} \implies \Theta(n^2)$ time.
     - **Auxiliary Space:** $\Theta(1)$ since comparisons are performed in-place with no extra arrays.

2. **Comparison with Baseline Sorting Algorithms:**
   - **Quick Sort:** Average time $\Theta(n \log n)$, auxiliary space $O(\log n)$ call stack.
   - **Insertion Sort:** Worst/Average time $\Theta(n^2)$, auxiliary space $O(1)$.

3. **Experimental vs. Theoretical Results:**
   - Experimental timings closely follow theoretical expectations: Brute Force and Insertion Sort display steep quadratic curves ($O(n^2)$).
   - Divide and Conquer and Quick Sort scale efficiently with near-linear curves ($O(n \log n)$), demonstrating practical viability for large input sizes.

## Problem 2: Fast Exponentiation
Compute $a^n$ for large values of $n$.

1. **Repeated Multiplication:** Linear iterative multiplication ($O(n)$).
2. **Recursive Exponentiation:** Piecewise recurrence computing $a^{n/2} \times a^{n/2}$ without caching ($O(n)$).
3. **Divide and Conquer:** Computes $a^{n/2}$ once and reuses the result ($O(\log n)$).

````python
# 1. Repeated Multiplication
def power_iterative(a, n):
    res = 1
    for _ in range(n):
        res *= a
    return res

# 2. Recursive Exponentiation (Naive Branching from PDF Relation)
def power_recursive_naive(a, n):
    if n == 0:
        return 1
    if n % 2 == 0:
        return power_recursive_naive(a, n // 2) * power_recursive_naive(a, n // 2)
    else:
        return a * power_recursive_naive(a, n - 1)

# 3. Divide and Conquer Exponentiation (Subproblem Reused)
def power_dc(a, n):
    if n == 0:
        return 1
    half = power_dc(a, n // 2)
    if n % 2 == 0:
        return half * half
    else:
        return a * half * half
````

````python
# Verification on Sample Input: a = 2, n = 10
a, n = 2, 10
p_iter = power_iterative(a, n)
p_naive = power_recursive_naive(a, n)
p_dc = power_dc(a, n)

print(f"Computing {a}^{n}:")
print(f"Repeated Multiplication : {p_iter}")
print(f"Naive Recursive         : {p_naive}")
print(f"Divide and Conquer      : {p_dc}")
````

````python
# Performance Benchmarking: Problem 2
n_values = [50, 100, 150, 200, 250, 300, 400, 500]
t_iter, t_naive, t_fast = [], [], []
a = 2

for n in n_values:
    t0 = time.perf_counter()
    power_iterative(a, n)
    t_iter.append(time.perf_counter() - t0)
    
    t0 = time.perf_counter()
    power_recursive_naive(a, n)
    t_naive.append(time.perf_counter() - t0)
    
    t0 = time.perf_counter()
    power_dc(a, n)
    t_fast.append(time.perf_counter() - t0)

# Plotting Input Size vs Execution Time
plt.figure(figsize=(9, 5))
plt.plot(n_values, t_iter, marker='o', label='Repeated Multiplication (O(n))')
plt.plot(n_values, t_naive, marker='s', label='Naive Recursive (O(n))')
plt.plot(n_values, t_fast, marker='^', label='Divide & Conquer (O(log n))')

plt.title('Problem 2: Exponent (n) vs Execution Time')
plt.xlabel('Exponent (n)')
plt.ylabel('Execution Time (seconds)')
plt.legend()
plt.grid(True)
plt.show()
````

### Problem 2: Theoretical Analysis & Lab Questions

1. **Recurrence Relation for Divide and Conquer:**
   $$T(n) = T(\lfloor n/2 \rfloor) + \Theta(1)$$
   - Applying Master Theorem ($a = 1, b = 2, d = 0$ where $\log_b a = \log_2 1 = 0 = d$):
   $$T(n) = \Theta(\log n)$$

2. **Time and Auxiliary Space Complexity:**
   | Approach | Time Complexity | Auxiliary Space Complexity |
   | :--- | :--- | :--- |
   | Repeated Multiplication | $\Theta(n)$ | $\Theta(1)$ |
   | Naive Recursive Exponentiation | $\Theta(n)$ | $\Theta(\log n)$ call stack |
   | Divide and Conquer Exponentiation | $\Theta(\log n)$ | $\Theta(\log n)$ call stack |

3. **Effect of Repeatedly Computing the Same Subproblem in the Recursive Approach:**
   - In naive recursive exponentiation, when $n$ is even, $a^{n/2}$ is evaluated twice in separate branches (`power_recursive_naive(a, n//2) * power_recursive_naive(a, n//2)`).
   - This recomputes identical subproblems repeatedly across recursion branches, causing the total function calls to scale linearly with $n$ (roughly $2n$) rather than logarithmically.
   - The Divide and Conquer approach eliminates redundant work by computing $a^{\lfloor n/2 \rfloor}$ once and reusing it, executing in only $\lfloor \log_2 n \rfloor$ recursive steps.

4. **Why Reducing Multiplications Significantly Affects Execution Time for Large $n$:**
   - For large $n$, intermediate powers become numbers with thousands or millions of bits.
   - Multiplying large numbers is not an $O(1)$ hardware operation; it takes $O(b \log b)$ or $O(b^{\log_2 3})$ bit operations.
   - Reducing multiplications from $n$ to $\log_2 n$ eliminates thousands of expensive arbitrary-precision multiplications, significantly speeding up execution.

5. **Iterative Divide and Conquer Implementation and Auxiliary Space:**
   - Divide and Conquer exponentiation can be implemented iteratively using binary exponentiation:
   ```python
   def power_dc_iterative(a, n):
       res = 1
       while n > 0:
           if n % 2 == 1:
               res *= a
           a *= a
           n //= 2
       return res
   ```
   - **Effect on Auxiliary Space:** Eliminates the recursion call stack entirely, reducing auxiliary space from $\Theta(\log n)$ to $\Theta(1)$.

6. **Paradigm Comparison: Problem 1 vs Problem 2:**
   - **Divide:**
     - *Problem 1:* Divides array into 2 halves of size $n/2$.
     - *Problem 2:* Reduces exponent $n$ into a single subproblem of size $n/2$.
   - **Conquer:**
     - *Problem 1:* Solves both subproblems recursively ($2T(n/2)$).
     - *Problem 2:* Solves only one subproblem ($1T(n/2)$).
   - **Combine:**
     - *Problem 1:* Combines sorted halves in linear time $\Theta(n)$ during the merge step.
     - *Problem 2:* Combines subproblem solutions in constant time $\Theta(1)$ with a single multiplication.
   - **Overall Complexity:** Problem 1 is $\Theta(n \log n)$; Problem 2 is $\Theta(\log n)$.

7. **Difference Between Theoretical Complexity and Experimentally Observed Time:**
   - **Theoretical Complexity ($O$-notation):** Measures asymptotic operation growth as $n \to \infty$, abstracting away hardware constants, cache hierarchies, and memory allocation.
   - **Experimental Execution Time:** Measures elapsed wall-clock time on a physical machine, subject to CPU frequency scaling, operating system scheduling, timer resolution, and Python bytecode interpretation overhead.
